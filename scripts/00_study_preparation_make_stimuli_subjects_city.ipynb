```python
import os, sys, random, re
import pandas as pd
import numpy as np
```

```python
spots_list = [  'ad_brunch',         
                'health_bingedrinking',
                'ad_burger',
                'health_smoking',
                'ad_coffee',
                'health_drugabuse',
                'ad_pizza',
                'health_buckleup',
                'ad_furniture',
                'health_textingdriving',
                'ad_hotel',
                'health_diabetes',
                'ad_lawyer',
                'health_HIV',
                'ad_smartphone',
                'health_sunprotection',
                'ad_fundraising',
                'health_handwashing', 
                'ad_travelagency' ,
                'health_vaccination' ,
                'ad_carinsurance' ,
                'health_healthydiet' ,
                'ad_cosmetics' ,
                'health_employee' ,
                'ad_moviepremier' ,
                'health_jogging' ,
                'ad_petstore' ,
                'health_getsleep' ,
                'ad_clothessale' ,
                'health_doctor' ,
                'ad_carpetcleaning' ,
                'health_caffeine' ,
                'ad_grocerystore' ,
                'health_greatlake' ,
                'ad_potteryclass' ,
                'health_mentalhealth' ,
                'ad_movingbusiness' ,
                'health_recycling' ,
                'ad_museum' ,
                'health_polarbears' ,
             ]       


#small_large_list = ['_small',
#                    '_large'] * int(len(spots_list)/2)

less_more_ag = ['moreatt_',
                'lessatt_'] * int(len(spots_list)/2)

more_less_ag = ['lessatt_',
                'moreatt_'] * int(len(spots_list)/2)


prefix = 'resources/billboards/'
postfix ='.bmp'

n_billboards = len(spots_list)
n_billboards
```

```python
# say 40 participants -- everybody gets 40 messages. ...
```

```python
subj_file_names = []
for i in (np.arange(40)+1).tolist():
     subj_file_names.append('sub' + (str(i).zfill(3)) + '_billboards_positions.csv')

subj_file_names[:6]
```

```python
sub_count = 0

for current_subj in range(40):
    print('-----')
    print(current_subj)
    print('-----')
    
    sub_count = sub_count +1
    print(sub_count)

    current_list  = []
    current_list_inv = []

    random.shuffle(spots_list)
    #random.shuffle(commercial_list)

    
    
    for counter in range(n_billboards):
        current_type          = less_more_ag[counter]
        current_type_inv      = more_less_ag[counter]
        current_spots_bb      = spots_list[counter] 
    
        current_list.append(prefix + current_type + current_spots_bb      + postfix)
    
        current_list_inv.append(prefix + current_type_inv + current_spots_bb   + postfix)
    
    #complicated/more than it has to be function to sort both lists the same way
    sorting_array = np.arange(40)

    #np.random.shuffle(sorting_array)
    print(sorting_array)
    arrayed_current_list = np.asarray(current_list)
    arrayed_current_list_sorted = arrayed_current_list[sorting_array]

    arrayed_current_list_inv = np.asarray(current_list_inv)
    arrayed_current_list_sorted_inv = arrayed_current_list_inv[sorting_array]


    
    
    df = pd.DataFrame()
    df['position'] = np.arange(40)+1
    df['billboard_name'] = arrayed_current_list_sorted
    df.to_csv('../data/00_study_preparation/stimuli_subject_files/' + subj_file_names[sub_count-1], index=False)
    print(arrayed_current_list_sorted[:1])
    print(subj_file_names[sub_count-1])
        
    print('---')
    sub_count = sub_count +1
    print(sub_count)
    df['billboard_name'] = arrayed_current_list_sorted_inv
    df.to_csv('../data/00_study_preparation/stimuli_subject_files/' + subj_file_names[sub_count-1], index=False)
    print(arrayed_current_list_sorted_inv[:1])
    print(subj_file_names[sub_count-1])
    
    print('-----')
```
