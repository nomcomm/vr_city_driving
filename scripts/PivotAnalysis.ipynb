```python
import os
import pandas as pd
```

```python
large_df = pd.read_csv('../data/04_data_analysis/00_all_data.csv')
large_df.head()
```

```python
grouped = large_df.groupby(['subject', 'billboard_attention_level', 'condition']).agg({'billboard#_fixated':'sum', 'billboard#_duration':'sum',  'recall':'sum', 'recognition':'sum'})
grouped.reset_index(inplace=True)
grouped
```

```python
high_low_att_fixbinarized = grouped.pivot(index='subject', columns='billboard_attention_level', values = ['billboard#_fixated', 'billboard#_duration', 'recall', 'recognition'] )
high_low_att_fixbinarized.columns = high_low_att_fixbinarized.columns.map('_'.join)
high_low_att_fixbinarized.to_csv('../data/04_data_analysis/billboad_fixation_duration_recall_recognition_att_high_att_low.csv')
high_low_att_fixbinarized.head()
```
