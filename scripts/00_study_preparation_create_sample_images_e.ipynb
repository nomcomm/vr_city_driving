```python
import PIL.Image
import os
from PIL import ImageDraw, ImageFont

os.chdir('../data/00_study_preparation/black_sample_images/')
```

```python
#ls
```

```python
factor_attention_list = ['moreattentiongrabbing', 'lessattentiongrabbing']
factor_size_list = ['small', 'large']
```

```python
for i in range(40):
    for f1 in range(len(factor_attention_list)):
        for f2 in range(len(factor_size_list)):
             current_name = 'topic#' + str(i+1) + '_' + factor_attention_list[f1] + '_' + factor_size_list[f2]
             print(current_name)

             if (factor_size_list[f2] == 'small'):
                image = PIL.Image.new('RGB', (200, 200))
                draw = PIL.ImageDraw.Draw(image)
                font = PIL.ImageFont.truetype("Arial.ttf", 10)
                draw.text((10, 70), current_name, fill='white', font = font)
                outfile_name = current_name + '.bmp' 
                image.save(outfile_name)
        
             elif (factor_size_list[f2] == 'large'):
                image = PIL.Image.new('RGB', (400, 400))
                draw = PIL.ImageDraw.Draw(image)
                font = PIL.ImageFont.truetype("Arial.ttf", 18)
                draw.text((50, 140), current_name, fill='white', font = font)
                outfile_name = current_name + '.bmp' 
                image.save(outfile_name)
```
